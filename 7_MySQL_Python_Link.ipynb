```python
# Use the newly created database
cursor.execute("USE MarceloDB")

# This SQL command creates a table named MarceloTable
# with columns for id, First_Name, Last_Name, Email, and Phone_Number. The id column is set as the primary key and will auto-increment with each new record.
cursor.execute("""
CREATE TABLE IF NOT EXISTS MarceloTable (
    id INT AUTO_INCREMENT PRIMARY KEY,
    First_Name VARCHAR(255),
    Last_Name VARCHAR(255),
    Email VARCHAR(255),
    Phone_Number VARCHAR(20)
)
""")

# Commit the changes to the database
conn.commit()

print("Table created successfully!")
```

```python
# This will insert multiple records into the MarceloTable using a single SQL command.
# The values to be inserted are provided as a list of tuples, where each tuple represents a record to be added to the table.
sql = """
INSERT INTO MarceloTable
(First_Name, Last_Name, Email, Phone_Number)
VALUES (%s, %s, %s, %s)
"""

values = [
    ("Marcelo", "Perez", "marcelo.perez@example.com", "123-456-7890"),
    ("John", "Doe", "john.doe@example.com", "098-765-4321"),
    ("Jane", "Smith", "jane.smith@example.com", "555-123-4567")
]

# executemany() method is used to execute the SQL command for each set of values in the list,
# allowing for efficient insertion of multiple records at once.
cursor.executemany(sql, values)

conn.commit()

print(f"{cursor.rowcount} rows inserted.")
```

```python
# This SQL commands will creates several tables named: EMPLOYEES, JOB_HISTORY, JOBS, DEPARTMENTS, and LOCATIONS with their respective columns and primary keys. 
# The IF NOT EXISTS clause ensures that the tables are only created if they do not already exist in the database.

cursor.execute("""
CREATE TABLE IF NOT EXISTS EMPLOYEES (
    EMP_ID CHAR(9) NOT NULL, 
    F_NAME VARCHAR(15) NOT NULL,
    L_NAME VARCHAR(15) NOT NULL,
    SSN CHAR(9),
    B_DATE DATE,
    SEX CHAR,
    ADDRESS VARCHAR(30),
    JOB_ID CHAR(9),
    SALARY DECIMAL(10,2),
    MANAGER_ID CHAR(9),
    DEP_ID CHAR(9) NOT NULL,
    PRIMARY KEY (EMP_ID)
    )""")
                            
cursor.execute("""
  CREATE TABLE IF NOT EXISTS JOB_HISTORY (
    EMPL_ID CHAR(9) NOT NULL, 
    START_DATE DATE,
    JOBS_ID CHAR(9) NOT NULL,
    DEPT_ID CHAR(9),
    PRIMARY KEY (EMPL_ID,JOBS_ID)
    )""")
 
cursor.execute("""
 CREATE TABLE IF NOT EXISTS JOBS (
    JOB_IDENT CHAR(9) NOT NULL, 
    JOB_TITLE VARCHAR(30),
    MIN_SALARY DECIMAL(10,2),
    MAX_SALARY DECIMAL(10,2),
    PRIMARY KEY (JOB_IDENT)
    )""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS DEPARTMENTS (
    DEPT_ID_DEP CHAR(9) NOT NULL, 
    DEP_NAME VARCHAR(15) ,
    MANAGER_ID CHAR(9),
    LOC_ID CHAR(9),
    PRIMARY KEY (DEPT_ID_DEP)
    )""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS LOCATIONS (
    LOCT_ID CHAR(9) NOT NULL,
    DEP_ID_LOC CHAR(9) NOT NULL,
    PRIMARY KEY (LOCT_ID,DEP_ID_LOC)
                            )""")

                          
```

```python
from mysql.connector import Error

try:
    cursor.execute("USE MarceloDB")
    
    # Disable foreign key checks to allow deletion in any order
    cursor.execute("SET FOREIGN_KEY_CHECKS = 0")

    tables_to_clear = [
        "JOB_HISTORY",
        "EMPLOYEES",
        "DEPARTMENTS",
        "JOBS",
        "LOCATIONS"
    ]

    for table in tables_to_clear:
        cursor.execute(f"DELETE FROM `{table}`")
        print(f"Cleared table: {table}")

    cursor.execute("SET FOREIGN_KEY_CHECKS = 1")
    conn.commit()
    print("All data deleted successfully.")

except Error as err:
    cursor.execute("SET FOREIGN_KEY_CHECKS = 1")
    conn.rollback()
    print(f"Error: {err}")
```

```python
# This code snippet is designed to load data from CSV files into corresponding MySQL tables
# Then it displays the contents of those tables in a formatted manner.    

import csv
from tabulate import tabulate

cursor.execute("USE MarceloDB")

# 1. List of tables and their corresponding SQL files
# Adjust these names to match your actual files and table names
data_files = {
    "Employees": "CSVfile/Employees.csv",
    "Departments": "CSVfile/Departments.csv",
    "Job_History": "CSVfile/JobsHistory.csv",
    "Jobs": "CSVfile/Jobs.csv",
    "Locations": "CSVfile/Locations.csv"
}
# 2. Loop through each table/file pair
for table_name, sql_file in data_files.items():
    print(f"\nLoading data for table: {table_name}...")
    
    try:
        with open(sql_file, 'r', encoding='utf-8', newline='') as file:
            rows = [row for row in csv.reader(file) if row]
        
        if rows:
            placeholders = ", ".join(["%s"] * len(rows[0]))
            cursor.executemany(
                f"INSERT INTO `{table_name}` VALUES ({placeholders})",
                rows
            )
        
        conn.commit()
        print(f"Successfully loaded data into {table_name}")
        
    except FileNotFoundError:
        print(f"Error: File '{sql_file}' not found.")
    except mysql.connector.Error as err:
        print(f"Error loading data into {table_name}: {err}")
        conn.rollback() # Undo changes if something went wrong

# 3. Now let's print the tables to verify
# 3. Now let's print the tables to verify
cursor.execute("USE MarceloDB")
cursor.execute("SHOW TABLES")
tables = cursor.fetchall()

for table_tuple in tables:
    table_name = table_tuple[0]
    print(f"\n{'='*50}")
    print(f"TABLE: {table_name}")
    print(f"{'='*50}")
    
    try:
        cursor.execute(f"SELECT * FROM `{table_name}` LIMIT 10") # Limit to 10 rows for preview
        rows = cursor.fetchall()
        columns = [col[0] for col in cursor.description]
        
        if rows:
            print(tabulate(rows, headers=columns, tablefmt="grid"))
        else:
            print("No records found.")
    except Exception as e:
        print(f"Error reading table {table_name}: {e}")
```

```python
# Define the query.
# select all employees whose address contains the letter 'S'
query = """
SELECT F_NAME, L_NAME, ADDRESS
FROM EMPLOYEES WHERE ADDRESS LIKE '%S%';
"""

# Execute the query
cursor.execute(query)

# Fetch all results
rows = cursor.fetchall()

# Get column names from the cursor description
columns = [col[0] for col in cursor.description]

# Print formatted output
print(tabulate(rows, headers=columns, tablefmt="grid"))
```

```python
# Define the query.

query = """
SELECT DEP_ID, COUNT(*) AS EMPLOYEE_COUNT, AVG(SALARY) AS AVERAGE_SALARY

FROM EMPLOYEES

GROUP BY DEP_ID

ORDER BY AVERAGE_SALARY;

"""

# Execute the query
cursor.execute(query)

# Fetch all results
rows = cursor.fetchall()

# Get column names from the cursor description
columns = [col[0] for col in cursor.description]

# Print formatted output
print(tabulate(rows, headers=columns, tablefmt="grid"))
```
